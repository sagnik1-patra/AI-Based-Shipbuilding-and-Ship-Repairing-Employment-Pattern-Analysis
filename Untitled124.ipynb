```python
import os
import json
import yaml
import pickle
import warnings

from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.base import clone
from sklearn.model_selection import (
    LeaveOneOut,
    StratifiedKFold,
    cross_val_predict
)

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)

from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
np.random.seed(SEED)
rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

# ============================================================
# 1. PROJECT DIRECTORY
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Shipbuilding and Ship Repairing Employment Pattern Analysis"
)

DATASET_PATH = BASE_DIR / (
    "10.3_Employment_in_Ship_Building_"
    "Ship_Repairing_Companies_As_on_31st_March_1996.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

def output(filename):
    return BASE_DIR / ("pso_" + filename)

def save_show(filename):
    plt.tight_layout()
    plt.savefig(
        output(filename),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

print("=" * 70)
print("PSO SHIPBUILDING EMPLOYMENT ANALYSIS")
print("=" * 70)
print("Dataset:", DATASET_PATH)
print("Output:", BASE_DIR)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

df.columns = (
    df.columns.astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("\nOriginal dataset shape:", df.shape)
print("\nDataset columns:")
print(df.columns.tolist())

print("\nDataset preview:")
display(df.head())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

rename_map = {
    "S. No.": "Serial_Number",
    "Category of Sectors": "Sector",
    "Name of the Company": "Company",
    "Officers": "Officers",
    "Supervisors": "Supervisors",
    "Workforce - Directly Involved in Operation": "Direct_Workers",
    "Workforce - Indirectly Involved in Operation": "Indirect_Workers",
    "Total": "Total_Employment"
}

df = df.rename(columns=rename_map)

required = [
    "Sector",
    "Company",
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}\n"
        f"Available: {df.columns.tolist()}"
    )

# ============================================================
# 4. CLEAN DATASET
# ============================================================

df["Sector"] = (
    df["Sector"]
    .ffill()
    .astype(str)
    .str.strip()
)

df["Company"] = (
    df["Company"]
    .astype(str)
    .str.strip()
)

# Remove aggregate rows if present.
df = df[
    ~df["Company"].str.fullmatch(
        r"(?i)\s*(grand\s+total|sub\s*total|total)\s*",
        na=False
    )
].copy()

numeric_cols = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

df = df.dropna(
    subset=["Total_Employment"]
).copy()

df = df[
    df["Total_Employment"] > 0
].copy()

df = df.reset_index(drop=True)

if len(df) < 8:
    raise ValueError(
        "Insufficient valid records for PSO evaluation."
    )

print("\nCleaned dataset shape:", df.shape)
print("\nMissing values:")
print(df.isna().sum())

# ============================================================
# 5. EMPLOYMENT CATEGORY CLASSIFICATION
# ============================================================

# Define descriptive employment categories using tertiles.
# These are not standardized industry thresholds.

q1, q2 = np.quantile(
    df["Total_Employment"],
    [1/3, 2/3]
)

def employment_category(total):
    if total <= q1:
        return "Low"
    elif total <= q2:
        return "Medium"
    return "High"

df["Employment_Category"] = (
    df["Total_Employment"]
    .apply(employment_category)
)

# Workforce composition features.
# Do not use raw total employment as a predictor because
# employment categories are derived from that total.

workforce_cols = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers"
]

for col in workforce_cols:
    df[col + "_Share"] = (
        df[col] / df["Total_Employment"]
    )

all_features = [
    "Sector",
    "Officers_Share",
    "Supervisors_Share",
    "Direct_Workers_Share",
    "Indirect_Workers_Share"
]

X = df[all_features].copy()

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df["Employment_Category"]
)

class_names = list(label_encoder.classes_)

n_classes = len(class_names)
n_features = len(all_features)

print("\nEmployment thresholds:")
print("Low <=", round(q1, 2))
print("Medium <=", round(q2, 2))
print("High >", round(q2, 2))

print("\nClass distribution:")
print(df["Employment_Category"].value_counts())

print("\nAvailable features:")
print(all_features)

# ============================================================
# 6. PREPROCESSING PIPELINE
# ============================================================

def create_preprocessor(selected_features):

    numeric_features = [
        feature
        for feature in selected_features
        if feature != "Sector"
    ]

    categorical_features = [
        feature
        for feature in selected_features
        if feature == "Sector"
    ]

    transformers = []

    if numeric_features:

        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append((
            "numeric",
            numeric_pipeline,
            numeric_features
        ))

    if categorical_features:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ])

        transformers.append((
            "categorical",
            categorical_pipeline,
            categorical_features
        ))

    return ColumnTransformer(transformers)

def make_pipeline(model, features):

    return Pipeline([
        (
            "preprocessor",
            create_preprocessor(features)
        ),
        (
            "classifier",
            clone(model)
        )
    ])

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {

    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=3,
        random_state=SEED
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=3,
        random_state=SEED
    ),

    "SVM": SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale"
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=3
    )
}

# ============================================================
# 8. PSO FITNESS FUNCTION
# ============================================================

# Binary PSO selects feature subsets.
#
# Each particle contains 0 or 1 for each feature.
# 1 = feature selected
# 0 = feature excluded
#
# Fitness = Weighted F1 - Feature Penalty
#
# Logistic Regression is used to evaluate each particle.

class_counts = np.bincount(y)

min_class_count = int(
    class_counts[class_counts > 0].min()
)

if min_class_count < 2:
    raise ValueError(
        "At least two records per class are required."
    )

fitness_cv = StratifiedKFold(
    n_splits=min(3, min_class_count),
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def mask_to_features(mask):

    return [
        all_features[i]
        for i in range(n_features)
        if mask[i] == 1
    ]

def fitness_function(mask):

    key = tuple(int(v) for v in mask)

    if key in fitness_cache:
        return fitness_cache[key]

    selected_features = mask_to_features(mask)

    if not selected_features:
        return -1.0

    model = LogisticRegression(
        max_iter=2000,
        random_state=SEED
    )

    pipeline = make_pipeline(
        model,
        selected_features
    )

    predictions = cross_val_predict(
        pipeline,
        X[selected_features],
        y,
        cv=fitness_cv,
        n_jobs=1
    )

    score = f1_score(
        y,
        predictions,
        average="weighted",
        zero_division=0
    )

    penalty = (
        0.02 *
        len(selected_features) /
        n_features
    )

    fitness = float(score - penalty)

    fitness_cache[key] = fitness

    return fitness

# ============================================================
# 9. BINARY PARTICLE SWARM OPTIMIZATION
# ============================================================

NUM_PARTICLES = 15
MAX_ITERATIONS = 20

INERTIA = 0.7
COGNITIVE = 1.5
SOCIAL = 1.5

# Initialize particle positions.
positions = rng.integers(
    0,
    2,
    size=(NUM_PARTICLES, n_features)
)

# Ensure no particle has an empty feature subset.
for i in range(NUM_PARTICLES):
    if not positions[i].any():
        positions[i, rng.integers(n_features)] = 1

# Include all features as a starting candidate.
positions[0] = np.ones(n_features, dtype=int)

# Initialize velocities.
velocities = rng.uniform(
    -1,
    1,
    size=(NUM_PARTICLES, n_features)
)

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    fitness_function(position)
    for position in positions
])

global_best_index = np.argmax(
    personal_best_scores
)

global_best_position = (
    personal_best_positions[global_best_index].copy()
)

global_best_score = float(
    personal_best_scores[global_best_index]
)

fitness_history = []
mean_fitness_history = []

print("\n" + "=" * 70)
print("PARTICLE SWARM OPTIMIZATION STARTED")
print("=" * 70)

for iteration in range(MAX_ITERATIONS):

    for i in range(NUM_PARTICLES):

        r1 = rng.random(n_features)
        r2 = rng.random(n_features)

        # Update velocity.
        velocities[i] = (
            INERTIA * velocities[i]
            +
            COGNITIVE * r1 * (
                personal_best_positions[i] -
                positions[i]
            )
            +
            SOCIAL * r2 * (
                global_best_position -
                positions[i]
            )
        )

        # Clip velocity to prevent numerical instability.
        velocities[i] = np.clip(
            velocities[i],
            -6,
            6
        )

        # Binary PSO sigmoid transformation.
        sigmoid = (
            1 / (1 + np.exp(-velocities[i]))
        )

        positions[i] = (
            rng.random(n_features) < sigmoid
        ).astype(int)

        # Ensure at least one selected feature.
        if not positions[i].any():
            positions[
                i,
                rng.integers(n_features)
            ] = 1

        current_score = fitness_function(
            positions[i]
        )

        # Update personal best.
        if current_score > personal_best_scores[i]:

            personal_best_scores[i] = current_score

            personal_best_positions[i] = (
                positions[i].copy()
            )

            # Update global best.
            if current_score > global_best_score:

                global_best_score = float(
                    current_score
                )

                global_best_position = (
                    positions[i].copy()
                )

    current_scores = [
        fitness_function(position)
        for position in positions
    ]

    fitness_history.append(
        global_best_score
    )

    mean_fitness_history.append(
        float(np.mean(current_scores))
    )

    print(
        f"Iteration {iteration + 1:02d}/{MAX_ITERATIONS} "
        f"| Best Fitness: {global_best_score:.4f}"
    )

selected_features = mask_to_features(
    global_best_position
)

print("\nPSO Optimization Completed")
print("Best Fitness:", global_best_score)
print("Selected Features:", selected_features)

# ============================================================
# 10. SAVE PSO SELECTED FEATURES
# ============================================================

selected_feature_data = {
    "algorithm": "Binary Particle Swarm Optimization",
    "best_fitness": float(global_best_score),
    "selected_features": selected_features,
    "selected_feature_count": len(selected_features),
    "total_feature_count": n_features,
    "particles": NUM_PARTICLES,
    "iterations": MAX_ITERATIONS
}

with open(
    output("selected_features.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        selected_feature_data,
        f,
        indent=4
    )

# ============================================================
# 11. PSO FITNESS CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, MAX_ITERATIONS + 1),
    fitness_history,
    marker="o",
    linewidth=2,
    color="royalblue",
    label="Global Best Fitness"
)

plt.plot(
    range(1, MAX_ITERATIONS + 1),
    mean_fitness_history,
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="Mean Particle Fitness"
)

plt.title(
    "PSO Fitness Convergence",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Iteration")
plt.ylabel("Fitness Score")
plt.legend()
plt.grid(alpha=0.3)

save_show("fitness_graph.png")

# ============================================================
# 12. PSO FEATURE SELECTION GRAPH
# ============================================================

feature_selection_df = pd.DataFrame({
    "Feature": all_features,
    "Selected": global_best_position.astype(int)
})

plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_selection_df,
    x="Selected",
    y="Feature",
    color="teal"
)

plt.title(
    "PSO Optimized Feature Selection",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Feature Selected (1 = Yes, 0 = No)")
plt.ylabel("Feature")
plt.xlim(0, 1.2)

save_show("feature_selection_graph.png")

feature_selection_df.to_csv(
    output("feature_selection.csv"),
    index=False
)

# ============================================================
# 13. TRAIN PSO-OPTIMIZED ML MODELS
# ============================================================

X_selected = X[selected_features].copy()

loo = LeaveOneOut()

results = []
model_predictions = {}

print("\n" + "=" * 70)
print("PSO OPTIMIZED MODEL EVALUATION")
print("=" * 70)

for model_name, model in models.items():

    pipeline = make_pipeline(
        model,
        selected_features
    )

    y_pred = cross_val_predict(
        pipeline,
        X_selected,
        y,
        cv=loo,
        n_jobs=1
    )

    accuracy = accuracy_score(
        y,
        y_pred
    )

    precision = precision_score(
        y,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y,
        y_pred,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    model_predictions[model_name] = y_pred

    print(f"\n{model_name}")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")

# ============================================================
# 14. SAVE PSO MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    output("model_comparison.csv"),
    index=False
)

print("\nPSO MODEL COMPARISON")
display(comparison_df)

best_model_name = comparison_df.iloc[0]["Model"]

best_predictions = model_predictions[
    best_model_name
]

print("\nBest PSO Model:", best_model_name)

# ============================================================
# 15. PSO ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    color="royalblue"
)

for bar in ax.patches:

    value = bar.get_height()

    ax.annotate(
        f"{value:.3f}",
        (
            bar.get_x() + bar.get_width()/2,
            value
        ),
        ha="center",
        va="bottom"
    )

plt.title(
    "PSO Optimized Model Accuracy Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("LOOCV Accuracy")
plt.ylim(0, 1.1)
plt.xticks(rotation=25, ha="right")

save_show("accuracy_graph.png")

# ============================================================
# 16. PSO CONFUSION MATRIX HEATMAP
# ============================================================

cm = confusion_matrix(
    y,
    best_predictions,
    labels=np.arange(n_classes)
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
    linewidths=0.5
)

plt.title(
    f"PSO Confusion Matrix - {best_model_name}",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

save_show("heatmap.png")

# ============================================================
# 17. PSO COMPARISON GRAPH
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score"
]

comparison_df.set_index("Model")[
    metrics
].plot(
    kind="bar",
    figsize=(12, 7),
    width=0.8
)

plt.title(
    "PSO Optimized Machine Learning Model Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Performance Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=25, ha="right")
plt.legend(title="Metrics")
plt.grid(axis="y", alpha=0.3)

save_show("comparison_graph.png")

# ============================================================
# 18. PSO RESULT CSV
# ============================================================

result_df = df[[
    "Company",
    "Sector",
    "Total_Employment",
    "Employment_Category"
]].copy()

result_df["Predicted_Category"] = (
    label_encoder.inverse_transform(
        best_predictions
    )
)

result_df["Correct_Prediction"] = (
    result_df["Employment_Category"]
    ==
    result_df["Predicted_Category"]
)

result_df["PSO_Model"] = best_model_name

result_df.to_csv(
    output("result.csv"),
    index=False
)

print("\nPSO CLASSIFICATION RESULTS")
display(result_df)

# ============================================================
# 19. PSO RESULT GRAPH
# ============================================================

actual_counts = (
    result_df["Employment_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

predicted_counts = (
    result_df["Predicted_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

result_counts = pd.DataFrame({
    "Actual": actual_counts,
    "PSO Predicted": predicted_counts
})

result_counts.plot(
    kind="bar",
    figsize=(10, 6),
    color=["steelblue", "darkorange"]
)

plt.title(
    "PSO Actual vs Predicted Employment Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Employment Category")
plt.ylabel("Number of Companies")
plt.xticks(rotation=0)
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("result_graph.png")

# ============================================================
# 20. PSO PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({
    "Company": df["Company"],
    "Sector": df["Sector"],
    "Actual_Total_Employment": df["Total_Employment"],
    "Actual_Category": label_encoder.inverse_transform(y),
    "Predicted_Category": label_encoder.inverse_transform(
        best_predictions
    ),
    "Prediction_Status": np.where(
        y == best_predictions,
        "Correct",
        "Incorrect"
    ),
    "Model": best_model_name
})

prediction_df.to_csv(
    output("prediction.csv"),
    index=False
)

print("\nPSO PREDICTION RESULTS")
display(prediction_df)

# ============================================================
# 21. PSO PREDICTION GRAPH
# ============================================================

ordered_classes = [
    category
    for category in ["Low", "Medium", "High"]
    if category in class_names
]

category_map = {
    name: i
    for i, name in enumerate(ordered_classes)
}

actual_numeric = (
    prediction_df["Actual_Category"]
    .map(category_map)
    .to_numpy()
)

predicted_numeric = (
    prediction_df["Predicted_Category"]
    .map(category_map)
    .to_numpy()
)

indices = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(14, 7))

plt.plot(
    indices,
    actual_numeric,
    marker="o",
    linewidth=2,
    label="Actual Category",
    color="royalblue"
)

plt.plot(
    indices,
    predicted_numeric,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted Category",
    color="darkorange"
)

plt.xticks(
    indices,
    prediction_df["Company"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    range(len(ordered_classes)),
    ordered_classes
)

plt.title(
    "PSO Actual vs Predicted Employment Categories",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Shipbuilding Companies")
plt.ylabel("Employment Category")
plt.legend()
plt.grid(alpha=0.3)

save_show("prediction_graph.png")

# ============================================================
# 22. PSO CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("PSO CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y,
        best_predictions,
        labels=np.arange(n_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 23. TRAIN FINAL PSO MODEL
# ============================================================

final_pipeline = make_pipeline(
    models[best_model_name],
    selected_features
)

final_pipeline.fit(
    X_selected,
    y
)

with open(
    output("shipbuilding_model.pkl"),
    "wb"
) as f:
    pickle.dump(
        final_pipeline,
        f
    )

with open(
    output("preprocessor.pkl"),
    "wb"
) as f:
    pickle.dump(
        final_pipeline.named_steps["preprocessor"],
        f
    )

with open(
    output("label_encoder.pkl"),
    "wb"
) as f:
    pickle.dump(
        label_encoder,
        f
    )

print("\nPSO optimized model saved.")

# ============================================================
# 24. EXPERIMENTAL NEURAL NETWORK
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

tf.random.set_seed(SEED)

X_nn = (
    final_pipeline.named_steps["preprocessor"]
    .transform(X_selected)
)

X_nn = np.asarray(
    X_nn,
    dtype=np.float32
)

nn_model = keras.Sequential([
    layers.Input(
        shape=(X_nn.shape[1],)
    ),

    layers.Dense(
        32,
        activation="relu"
    ),

    layers.Dropout(0.2),

    layers.Dense(
        16,
        activation="relu"
    ),

    layers.Dense(
        n_classes,
        activation="softmax"
    )
])

nn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = nn_model.fit(
    X_nn,
    y,
    epochs=80,
    batch_size=4,
    verbose=0
)

nn_model.save(
    str(output("shipbuilding_model.h5"))
)

pd.DataFrame(
    history.history
).to_csv(
    output("neural_network_history.csv"),
    index=False
)

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    linewidth=2,
    label="Training Accuracy"
)

plt.title(
    "PSO Feature-Selected Neural Network Training",
    fontweight="bold"
)

plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(alpha=0.3)

save_show("neural_network_accuracy.png")

# ============================================================
# 25. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based Shipbuilding and Ship Repairing "
        "Employment Pattern Analysis"
    ),

    "optimization_algorithm": (
        "Binary Particle Swarm Optimization"
    ),

    "dataset": DATASET_PATH.name,

    "historical_year": 1996,

    "target": "Employment_Category",

    "selected_features": selected_features,

    "best_model": best_model_name,

    "validation": "Leave-One-Out",

    "pso_parameters": {
        "particles": NUM_PARTICLES,
        "iterations": MAX_ITERATIONS,
        "inertia": INERTIA,
        "cognitive": COGNITIVE,
        "social": SOCIAL
    },

    "thresholds": {
        "low_max": float(q1),
        "medium_max": float(q2)
    },

    "random_state": SEED
}

with open(
    output("model_config.yaml"),
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 26. SAVE JSON METADATA
# ============================================================

best_row = comparison_df.iloc[0]

metadata = {
    "project": config["project"],

    "algorithm": "Binary PSO",

    "created_at": datetime.now().isoformat(),

    "dataset_records": int(len(df)),

    "selected_features": selected_features,

    "best_fitness": float(global_best_score),

    "best_model": best_model_name,

    "accuracy": float(best_row["Accuracy"]),

    "precision": float(best_row["Precision"]),

    "recall": float(best_row["Recall"]),

    "f1_score": float(best_row["F1_Score"]),

    "neural_network_training_accuracy": float(
        history.history["accuracy"][-1]
    ),

    "evaluation_method": "LOOCV",

    "limitations": [
        "Small historical dataset.",
        "PSO feature selection uses the full dataset.",
        "Reported LOOCV metrics are not nested-CV estimates.",
        "Feature selection can produce optimistic performance.",
        "Neural network accuracy is training accuracy only.",
        "Results are exploratory and not production forecasts."
    ]
}

with open(
    output("model_metadata.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4
    )

# ============================================================
# 27. VERIFY OUTPUT FILES
# ============================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "fitness_graph.png",
    "feature_selection_graph.png",
    "feature_selection.csv",
    "model_comparison.csv",
    "shipbuilding_model.pkl",
    "shipbuilding_model.h5",
    "preprocessor.pkl",
    "label_encoder.pkl",
    "model_config.yaml",
    "model_metadata.json",
    "selected_features.json",
    "neural_network_history.csv",
    "neural_network_accuracy.png"
]

print("\n" + "=" * 70)
print("PSO OUTPUT VERIFICATION")
print("=" * 70)

for filename in output_files:

    filepath = output(filename)

    if filepath.exists():

        size_kb = (
            filepath.stat().st_size / 1024
        )

        print(
            f"[SAVED] pso_{filename} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(
            f"[MISSING] pso_{filename}"
        )

print("\n" + "=" * 70)
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nBest PSO model:", best_model_name)
print("Selected features:", selected_features)
print("Best PSO fitness:", round(global_best_score, 4))

print("\nAll results saved in:")
print(BASE_DIR)
```
