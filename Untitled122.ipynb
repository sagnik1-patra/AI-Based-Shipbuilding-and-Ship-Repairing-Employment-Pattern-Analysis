```python
import os
import json
import yaml
import pickle
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    LeaveOneOut,
    cross_val_predict
)
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

warnings.filterwarnings("ignore", category=FutureWarning)

sns.set_theme(style="whitegrid")

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# ============================================================
# 1. PROJECT PATH
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Shipbuilding and Ship Repairing Employment Pattern Analysis"
)

DATASET_PATH = BASE_DIR / (
    "10.3_Employment_in_Ship_Building_"
    "Ship_Repairing_Companies_As_on_31st_March_1996.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 70)
print("SHIPBUILDING EMPLOYMENT ANALYSIS")
print("=" * 70)
print("Dataset:", DATASET_PATH)
print("Output folder:", BASE_DIR)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

df.columns = (
    df.columns.astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("\nDataset shape:", df.shape)
print("\nAvailable columns:")
print(df.columns.tolist())

print("\nDataset preview:")
display(df.head())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

rename_columns = {
    "S. No.": "Serial_Number",
    "Category of Sectors": "Sector",
    "Name of the Company": "Company",
    "Officers": "Officers",
    "Supervisors": "Supervisors",
    "Workforce - Directly Involved in Operation": "Direct_Workers",
    "Workforce - Indirectly Involved in Operation": "Indirect_Workers",
    "Total": "Total_Employment"
}

df = df.rename(columns=rename_columns)

required = [
    "Sector",
    "Company",
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Missing columns: {missing}\n"
        f"Available: {df.columns.tolist()}"
    )

# ============================================================
# 4. DATA CLEANING
# ============================================================

df["Sector"] = (
    df["Sector"]
    .ffill()
    .astype(str)
    .str.strip()
)

df["Company"] = (
    df["Company"]
    .astype(str)
    .str.strip()
)

# Remove total/subtotal rows if present
df = df[
    ~df["Company"].str.contains(
        r"^\s*(grand\s+total|sub\s*total|total)\s*$",
        case=False,
        regex=True,
        na=False
    )
].copy()

numeric_columns = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col].astype(str)
        .str.replace(",", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

df = df.dropna(
    subset=["Total_Employment"]
).copy()

df = df[
    df["Total_Employment"] > 0
].copy()

df = df.reset_index(drop=True)

if len(df) < 6:
    raise ValueError(
        "Insufficient valid company records for evaluation."
    )

print("\nCleaned dataset shape:", df.shape)
print("\nMissing values:")
print(df.isna().sum())

# ============================================================
# 5. FEATURE ENGINEERING
# ============================================================

# Employment categories are based on dataset tertiles.
# These are descriptive categories, not industry standards.

q1, q2 = np.quantile(
    df["Total_Employment"],
    [1 / 3, 2 / 3]
)

def classify_employment(value):
    if value <= q1:
        return "Low"
    elif value <= q2:
        return "Medium"
    return "High"

df["Employment_Category"] = (
    df["Total_Employment"]
    .apply(classify_employment)
)

# Use workforce composition shares instead of raw totals.
# The target is derived from total employment.
# Using the total or its raw component counts would cause
# serious target leakage.

workforce_columns = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers"
]

for col in workforce_columns:
    df[col + "_Share"] = (
        df[col] / df["Total_Employment"]
    )

feature_columns = [
    "Sector",
    "Officers_Share",
    "Supervisors_Share",
    "Direct_Workers_Share",
    "Indirect_Workers_Share"
]

X = df[feature_columns].copy()

encoder = LabelEncoder()
y = encoder.fit_transform(
    df["Employment_Category"]
)

class_names = list(encoder.classes_)
n_classes = len(class_names)

if n_classes < 2:
    raise ValueError(
        "At least two employment categories are required."
    )

print("\nEmployment thresholds:")
print(f"Low: <= {q1:.2f}")
print(f"Medium: > {q1:.2f} and <= {q2:.2f}")
print(f"High: > {q2:.2f}")

print("\nEmployment categories:")
print(df["Employment_Category"].value_counts())

# ============================================================
# 6. PREPROCESSING
# ============================================================

numeric_features = [
    "Officers_Share",
    "Supervisors_Share",
    "Direct_Workers_Share",
    "Indirect_Workers_Share"
]

categorical_features = ["Sector"]

def make_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ])

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=RANDOM_STATE
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=3,
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=3,
        random_state=RANDOM_STATE
    ),

    "SVM": SVC(
        kernel="rbf",
        C=1.0
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=3
    )
}

# ============================================================
# 8. LEAVE-ONE-OUT CROSS VALIDATION
# ============================================================

# Small datasets should not be evaluated with a single
# random train/test split.
#
# Each LOOCV iteration trains on n-1 records and predicts
# the remaining record.
#
# Preprocessing is fitted separately in every fold.

loo = LeaveOneOut()

evaluation_results = []
model_predictions = {}

print("\n" + "=" * 70)
print("MODEL TRAINING AND EVALUATION")
print("=" * 70)

for name, model in models.items():

    pipeline = Pipeline([
        (
            "preprocessor",
            make_preprocessor()
        ),
        (
            "classifier",
            model
        )
    ])

    pred = cross_val_predict(
        pipeline,
        X,
        y,
        cv=loo,
        n_jobs=1
    )

    accuracy = accuracy_score(y, pred)

    precision = precision_score(
        y,
        pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y,
        pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y,
        pred,
        average="weighted",
        zero_division=0
    )

    evaluation_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    model_predictions[name] = pred

    print(f"\n{name}")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")

# ============================================================
# 9. MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(
    evaluation_results
)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    BASE_DIR / "model_comparison.csv",
    index=False
)

print("\nMODEL COMPARISON")
display(comparison_df)

best_model_name = comparison_df.iloc[0]["Model"]

best_predictions = model_predictions[
    best_model_name
]

print("\nSelected model:", best_model_name)

# ============================================================
# 10. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    color="royalblue"
)

for bar in ax.patches:
    height = bar.get_height()

    ax.annotate(
        f"{height:.3f}",
        (
            bar.get_x() + bar.get_width() / 2,
            height
        ),
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.title(
    "Accuracy Comparison of Machine Learning Models",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("LOOCV Accuracy")
plt.ylim(0, 1.10)
plt.xticks(rotation=25, ha="right")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: accuracy_graph.png")

# ============================================================
# 11. CONFUSION MATRIX HEATMAP
# ============================================================

cm = confusion_matrix(
    y,
    best_predictions,
    labels=np.arange(n_classes)
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
    linewidths=0.5,
    linecolor="white"
)

plt.title(
    f"Confusion Matrix - {best_model_name}",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Predicted Employment Category")
plt.ylabel("Actual Employment Category")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: heatmap.png")

# ============================================================
# 12. COMPARISON GRAPH
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score"
]

ax = comparison_df.set_index("Model")[
    metrics
].plot(
    kind="bar",
    figsize=(13, 7),
    width=0.8
)

plt.title(
    "Comparison of Machine Learning Performance Metrics",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Performance Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=25, ha="right")
plt.legend(title="Metrics")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: comparison_graph.png")

# ============================================================
# 13. RESULT CSV
# ============================================================

# These are out-of-fold predictions.
# Every record is predicted by a model trained without it.

result_df = df[[
    "Company",
    "Sector",
    "Total_Employment",
    "Employment_Category"
]].copy()

result_df["Predicted_Category"] = (
    encoder.inverse_transform(
        best_predictions
    )
)

result_df["Prediction_Correct"] = (
    result_df["Employment_Category"]
    ==
    result_df["Predicted_Category"]
)

result_df["Model"] = best_model_name

result_df["Evaluation_Method"] = "LOOCV"

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print("\nCLASSIFICATION RESULTS")
display(result_df)

print("Saved: result.csv")

# ============================================================
# 14. RESULT GRAPH
# ============================================================

# Compare actual and predicted category counts.

actual_counts = (
    result_df["Employment_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

predicted_counts = (
    result_df["Predicted_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

result_counts = pd.DataFrame({
    "Actual": actual_counts,
    "Predicted": predicted_counts
})

ax = result_counts.plot(
    kind="bar",
    figsize=(10, 6),
    color=["steelblue", "orange"],
    width=0.7
)

plt.title(
    "Actual vs Predicted Employment Category Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Employment Category")
plt.ylabel("Number of Companies")
plt.xticks(rotation=0)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: result_graph.png")

# ============================================================
# 15. PREDICTION CSV
# ============================================================

# The prediction CSV contains company-level
# out-of-fold predictions from the best model.

prediction_df = pd.DataFrame({
    "Company": df["Company"],
    "Sector": df["Sector"],
    "Actual_Total_Employment": df["Total_Employment"],
    "Actual_Category": encoder.inverse_transform(y),
    "Predicted_Category": encoder.inverse_transform(
        best_predictions
    ),
    "Prediction_Status": np.where(
        y == best_predictions,
        "Correct",
        "Incorrect"
    ),
    "Model": best_model_name
})

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\nPREDICTION RESULTS")
display(prediction_df)

print("Saved: prediction.csv")

# ============================================================
# 16. PREDICTION GRAPH
# ============================================================

# Categories are encoded as Low=0, Medium=1, High=2
# according to an explicit ordering.

ordered_classes = [
    c for c in ["Low", "Medium", "High"]
    if c in class_names
]

category_to_number = {
    name: i
    for i, name in enumerate(ordered_classes)
}

actual_numeric = (
    prediction_df["Actual_Category"]
    .map(category_to_number)
    .to_numpy()
)

predicted_numeric = (
    prediction_df["Predicted_Category"]
    .map(category_to_number)
    .to_numpy()
)

company_indices = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(14, 7))

plt.plot(
    company_indices,
    actual_numeric,
    marker="o",
    linestyle="-",
    linewidth=2,
    label="Actual Category",
    color="royalblue"
)

plt.plot(
    company_indices,
    predicted_numeric,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Predicted Category",
    color="darkorange"
)

plt.yticks(
    range(len(ordered_classes)),
    ordered_classes
)

plt.xticks(
    company_indices,
    prediction_df["Company"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.title(
    "Actual vs Predicted Employment Classification",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Shipbuilding Companies")
plt.ylabel("Employment Category")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved: prediction_graph.png")

# ============================================================
# 17. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y,
        best_predictions,
        labels=np.arange(n_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 18. TRAIN FINAL MODEL ON ALL RECORDS
# ============================================================

# The final model is trained on all available records.
# The graphs above use LOOCV predictions instead of
# training-set predictions.

final_pipeline = Pipeline([
    ("preprocessor", make_preprocessor()),
    ("classifier", models[best_model_name])
])

final_pipeline.fit(X, y)

with open(
    BASE_DIR / "shipbuilding_model.pkl",
    "wb"
) as f:
    pickle.dump(final_pipeline, f)

with open(
    BASE_DIR / "preprocessor.pkl",
    "wb"
) as f:
    pickle.dump(
        final_pipeline.named_steps["preprocessor"],
        f
    )

with open(
    BASE_DIR / "label_encoder.pkl",
    "wb"
) as f:
    pickle.dump(encoder, f)

# ============================================================
# 19. SAVE YAML CONFIG
# ============================================================

config = {
    "project": (
        "AI-Based Shipbuilding and Ship Repairing "
        "Employment Pattern Analysis"
    ),
    "dataset": DATASET_PATH.name,
    "historical_year": 1996,
    "target": "Employment_Category",
    "features": feature_columns,
    "classes": class_names,
    "best_model": best_model_name,
    "validation": "Leave-One-Out Cross Validation",
    "thresholds": {
        "low_max": float(q1),
        "medium_max": float(q2)
    },
    "random_state": RANDOM_STATE
}

with open(
    BASE_DIR / "model_config.yaml",
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 20. SAVE JSON METADATA
# ============================================================

best_metrics = comparison_df.iloc[0]

metadata = {
    "project": config["project"],
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "best_model": best_model_name,
    "accuracy": float(best_metrics["Accuracy"]),
    "precision": float(best_metrics["Precision"]),
    "recall": float(best_metrics["Recall"]),
    "f1_score": float(best_metrics["F1_Score"]),
    "validation": "LOOCV",
    "categories": class_names,
    "note": (
        "Historical small-sample exploratory classification. "
        "Model selection and evaluation use the same LOOCV "
        "results, so selected-model performance can be optimistic."
    )
}

with open(
    BASE_DIR / "model_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4
    )

# ============================================================
# 21. VERIFY ALL OUTPUT FILES
# ============================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "model_comparison.csv",
    "shipbuilding_model.pkl",
    "preprocessor.pkl",
    "label_encoder.pkl",
    "model_config.yaml",
    "model_metadata.json"
]

print("\n" + "=" * 70)
print("OUTPUT FILE VERIFICATION")
print("=" * 70)

for filename in output_files:

    filepath = BASE_DIR / filename

    if filepath.exists():
        size = filepath.stat().st_size / 1024

        print(
            f"[SAVED] {filename} ({size:.2f} KB)"
        )
    else:
        print(f"[MISSING] {filename}")

print("\n" + "=" * 70)
print("PROJECT EXECUTION COMPLETED")
print("=" * 70)

print("\nAll results saved in:")
print(BASE_DIR)
```
