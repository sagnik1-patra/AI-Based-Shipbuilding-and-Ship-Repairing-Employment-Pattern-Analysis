```python
import os
import json
import yaml
import pickle
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.base import clone
from sklearn.model_selection import (
    LeaveOneOut,
    StratifiedKFold,
    cross_val_predict
)
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

# ============================================================
# 1. PROJECT PATH
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Shipbuilding and Ship Repairing Employment Pattern Analysis"
)

DATASET_PATH = BASE_DIR / (
    "10.3_Employment_in_Ship_Building_"
    "Ship_Repairing_Companies_As_on_31st_March_1996.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

def output(filename):
    return BASE_DIR / ("ais_" + filename)

def save_show(filename):
    plt.tight_layout()
    plt.savefig(
        output(filename),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

print("=" * 70)
print("AIS SHIPBUILDING EMPLOYMENT ANALYSIS")
print("=" * 70)
print("Dataset:", DATASET_PATH)
print("Output:", BASE_DIR)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

df.columns = (
    df.columns.astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("\nOriginal shape:", df.shape)
print("\nOriginal columns:")
print(df.columns.tolist())

display(df.head())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

rename_map = {
    "S. No.": "Serial_Number",
    "Category of Sectors": "Sector",
    "Name of the Company": "Company",
    "Officers": "Officers",
    "Supervisors": "Supervisors",
    "Workforce - Directly Involved in Operation": "Direct_Workers",
    "Workforce - Indirectly Involved in Operation": "Indirect_Workers",
    "Total": "Total_Employment"
}

df = df.rename(columns=rename_map)

required = [
    "Sector",
    "Company",
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# ============================================================
# 4. CLEAN DATA
# ============================================================

df["Sector"] = (
    df["Sector"]
    .ffill()
    .astype(str)
    .str.strip()
)

df["Company"] = (
    df["Company"]
    .astype(str)
    .str.strip()
)

# Exclude aggregate rows.
df = df[
    ~df["Company"].str.fullmatch(
        r"(?i)\s*(grand\s+total|sub\s*total|total)\s*",
        na=False
    )
].copy()

numeric_cols = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

df = df.dropna(
    subset=["Total_Employment"]
).copy()

df = df[
    df["Total_Employment"] > 0
].copy()

df = df.reset_index(drop=True)

if len(df) < 8:
    raise ValueError(
        "Insufficient valid records for AIS evaluation."
    )

print("\nCleaned shape:", df.shape)
print("\nMissing values:")
print(df.isna().sum())

# ============================================================
# 5. EMPLOYMENT CLASSIFICATION
# ============================================================

# Descriptive size categories from historical dataset.
# Thresholds are not standardized industry boundaries.

q1, q2 = np.quantile(
    df["Total_Employment"],
    [1/3, 2/3]
)

def categorize(total):
    if total <= q1:
        return "Low"
    elif total <= q2:
        return "Medium"
    return "High"

df["Employment_Category"] = (
    df["Total_Employment"].apply(categorize)
)

# Workforce composition features.
# Avoid direct use of total employment or raw worker counts
# when predicting a target derived from total employment.

workforce_cols = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers"
]

for col in workforce_cols:
    df[col + "_Share"] = (
        df[col] / df["Total_Employment"]
    )

all_features = [
    "Sector",
    "Officers_Share",
    "Supervisors_Share",
    "Direct_Workers_Share",
    "Indirect_Workers_Share"
]

X = df[all_features].copy()

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df["Employment_Category"]
)

class_names = list(label_encoder.classes_)

print("\nClass distribution:")
print(df["Employment_Category"].value_counts())

print("\nAvailable features:")
print(all_features)

# ============================================================
# 6. PREPROCESSOR
# ============================================================

def create_preprocessor(selected_features):

    numeric_features = [
        col for col in selected_features
        if col != "Sector"
    ]

    categorical_features = [
        col for col in selected_features
        if col == "Sector"
    ]

    transformers = []

    if numeric_features:
        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append((
            "numeric",
            numeric_pipeline,
            numeric_features
        ))

    if categorical_features:
        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ])

        transformers.append((
            "categorical",
            categorical_pipeline,
            categorical_features
        ))

    return ColumnTransformer(transformers)

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=3,
        random_state=SEED
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=80,
        max_depth=3,
        random_state=SEED
    ),

    "SVM": SVC(
        kernel="rbf",
        C=1.0
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=3
    )
}

# ============================================================
# 8. AIS FITNESS FUNCTION
# ============================================================

# AIS uses Logistic Regression as a feature-selection
# fitness evaluator.
#
# Fitness = weighted F1 - feature-count penalty.
#
# Preprocessing is performed inside each CV fold.
# The final selected feature subset is evaluated separately.

feature_count = len(all_features)

def mask_to_features(mask):
    return [
        all_features[i]
        for i in range(feature_count)
        if mask[i] == 1
    ]

def make_pipeline(model, features):
    return Pipeline([
        (
            "preprocessor",
            create_preprocessor(features)
        ),
        (
            "classifier",
            clone(model)
        )
    ])

# A separate stratified CV is used for AIS fitness.
# Leave-one-out is used later for model comparisons.

class_counts = np.bincount(y)
min_class_count = int(
    class_counts[class_counts > 0].min()
)

if min_class_count < 2:
    raise ValueError(
        "At least two records per class are required."
    )

n_splits = min(3, min_class_count)

fitness_cv = StratifiedKFold(
    n_splits=n_splits,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def calculate_fitness(mask):

    key = tuple(int(v) for v in mask)

    if key in fitness_cache:
        return fitness_cache[key]

    selected = mask_to_features(mask)

    if not selected:
        return -1.0

    pipeline = make_pipeline(
        LogisticRegression(
            max_iter=2000,
            random_state=SEED
        ),
        selected
    )

    pred = cross_val_predict(
        pipeline,
        X[selected],
        y,
        cv=fitness_cv,
        n_jobs=1
    )

    f1 = f1_score(
        y,
        pred,
        average="weighted",
        zero_division=0
    )

    feature_penalty = (
        0.02 * len(selected) / feature_count
    )

    fitness = float(f1 - feature_penalty)

    fitness_cache[key] = fitness

    return fitness

# ============================================================
# 9. ARTIFICIAL IMMUNE SYSTEM
# CLONAL SELECTION ALGORITHM
# ============================================================

POPULATION_SIZE = 12
GENERATIONS = 15
CLONE_FACTOR = 4
REPLACEMENT_RATE = 0.25

def random_antibody():
    mask = rng.integers(
        0, 2,
        size=feature_count,
        dtype=int
    )

    if not mask.any():
        mask[rng.integers(feature_count)] = 1

    return mask

def mutate_antibody(mask, mutation_rate):

    mutated = mask.copy()

    for i in range(len(mutated)):
        if rng.random() < mutation_rate:
            mutated[i] = 1 - mutated[i]

    if not mutated.any():
        mutated[rng.integers(len(mutated))] = 1

    return mutated

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

# Include the full feature set.
population[0] = np.ones(
    feature_count,
    dtype=int
)

best_mask = None
best_fitness = -np.inf

fitness_history = []
mean_fitness_history = []

print("\n" + "=" * 70)
print("AIS CLONAL SELECTION OPTIMIZATION")
print("=" * 70)

for generation in range(GENERATIONS):

    scored = [
        (calculate_fitness(mask), mask.copy())
        for mask in population
    ]

    scored.sort(
        key=lambda item: item[0],
        reverse=True
    )

    generation_best_fitness = scored[0][0]

    if generation_best_fitness > best_fitness:
        best_fitness = generation_best_fitness
        best_mask = scored[0][1].copy()

    fitness_history.append(best_fitness)

    mean_fitness_history.append(
        float(np.mean([s[0] for s in scored]))
    )

    print(
        f"Generation {generation + 1:02d}/{GENERATIONS} "
        f"| Best Fitness: {best_fitness:.4f}"
    )

    # Select high-affinity antibodies.
    selected_count = max(
        2,
        POPULATION_SIZE // 2
    )

    selected_antibodies = scored[:selected_count]

    clones = []

    for rank, (fitness, antibody) in enumerate(
        selected_antibodies
    ):

        number_of_clones = max(
            1,
            CLONE_FACTOR * (
                selected_count - rank
            )
        )

        # Higher-affinity antibodies mutate less.
        mutation_rate = min(
            0.5,
            0.08 + 0.05 * rank
        )

        for _ in range(number_of_clones):

            new_clone = mutate_antibody(
                antibody,
                mutation_rate
            )

            clones.append(new_clone)

    clone_scores = [
        (calculate_fitness(mask), mask.copy())
        for mask in clones
    ]

    combined = scored + clone_scores

    combined.sort(
        key=lambda item: item[0],
        reverse=True
    )

    # Elitism: preserve best solutions.
    keep_count = max(
        1,
        int(
            POPULATION_SIZE * (
                1 - REPLACEMENT_RATE
            )
        )
    )

    new_population = [
        item[1].copy()
        for item in combined[:keep_count]
    ]

    # Introduce new random antibodies.
    while len(new_population) < POPULATION_SIZE:
        new_population.append(
            random_antibody()
        )

    population = new_population

selected_features = mask_to_features(best_mask)

print("\nAIS Optimization Completed")
print("Best Fitness:", best_fitness)
print("Selected Features:", selected_features)

# ============================================================
# 10. SAVE SELECTED FEATURES
# ============================================================

selected_feature_data = {
    "algorithm": "AIS Clonal Selection",
    "best_fitness": float(best_fitness),
    "selected_features": selected_features,
    "selected_feature_count": len(selected_features),
    "total_feature_count": feature_count,
    "generations": GENERATIONS,
    "population_size": POPULATION_SIZE
}

with open(
    output("selected_features.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        selected_feature_data,
        f,
        indent=4
    )

# ============================================================
# 11. AIS FITNESS CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, GENERATIONS + 1),
    fitness_history,
    marker="o",
    linewidth=2,
    label="Best Fitness"
)

plt.plot(
    range(1, GENERATIONS + 1),
    mean_fitness_history,
    linestyle="--",
    linewidth=2,
    label="Mean Population Fitness"
)

plt.title(
    "AIS Clonal Selection Fitness Convergence",
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Fitness")
plt.legend()
plt.grid(alpha=0.3)

save_show("fitness_graph.png")

# ============================================================
# 12. AIS FEATURE SELECTION GRAPH
# ============================================================

feature_selection_df = pd.DataFrame({
    "Feature": all_features,
    "Selected": best_mask.astype(int)
})

plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_selection_df,
    x="Selected",
    y="Feature",
    color="teal"
)

plt.title(
    "AIS Optimized Feature Selection",
    fontweight="bold"
)

plt.xlabel("Selected (1 = Yes, 0 = No)")
plt.ylabel("Features")
plt.xlim(0, 1.2)

save_show("feature_selection_graph.png")

feature_selection_df.to_csv(
    output("feature_selection.csv"),
    index=False
)

# ============================================================
# 13. TRAIN AIS-OPTIMIZED MODELS
# ============================================================

X_selected = X[selected_features].copy()

loo = LeaveOneOut()

results = []
predictions = {}

print("\n" + "=" * 70)
print("AIS OPTIMIZED MODEL EVALUATION")
print("=" * 70)

for model_name, model in models.items():

    pipeline = make_pipeline(
        model,
        selected_features
    )

    y_pred = cross_val_predict(
        pipeline,
        X_selected,
        y,
        cv=loo,
        n_jobs=1
    )

    accuracy = accuracy_score(y, y_pred)

    precision = precision_score(
        y,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y,
        y_pred,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    predictions[model_name] = y_pred

    print(f"\n{model_name}")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")

# ============================================================
# 14. MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    output("model_comparison.csv"),
    index=False
)

print("\nAIS MODEL COMPARISON")
display(comparison_df)

best_model_name = comparison_df.iloc[0]["Model"]
best_predictions = predictions[best_model_name]

print("\nBest AIS Model:", best_model_name)

# ============================================================
# 15. AIS ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    color="royalblue"
)

for bar in ax.patches:

    value = bar.get_height()

    ax.annotate(
        f"{value:.3f}",
        (
            bar.get_x() + bar.get_width()/2,
            value
        ),
        ha="center",
        va="bottom"
    )

plt.title(
    "AIS Optimized Model Accuracy Comparison",
    fontweight="bold"
)

plt.xlabel("Machine Learning Model")
plt.ylabel("LOOCV Accuracy")
plt.ylim(0, 1.1)
plt.xticks(rotation=25, ha="right")

save_show("accuracy_graph.png")

# ============================================================
# 16. AIS CONFUSION MATRIX HEATMAP
# ============================================================

cm = confusion_matrix(
    y,
    best_predictions,
    labels=np.arange(len(class_names))
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
    linewidths=0.5
)

plt.title(
    f"AIS Confusion Matrix - {best_model_name}",
    fontweight="bold"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

save_show("heatmap.png")

# ============================================================
# 17. AIS COMPARISON GRAPH
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score"
]

comparison_df.set_index("Model")[
    metrics
].plot(
    kind="bar",
    figsize=(12, 7),
    width=0.8
)

plt.title(
    "AIS Optimized Machine Learning Model Comparison",
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Performance Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=25, ha="right")
plt.legend(title="Metrics")
plt.grid(axis="y", alpha=0.3)

save_show("comparison_graph.png")

# ============================================================
# 18. AIS RESULT CSV
# ============================================================

result_df = df[[
    "Company",
    "Sector",
    "Total_Employment",
    "Employment_Category"
]].copy()

result_df["Predicted_Category"] = (
    label_encoder.inverse_transform(
        best_predictions
    )
)

result_df["Correct_Prediction"] = (
    result_df["Employment_Category"]
    ==
    result_df["Predicted_Category"]
)

result_df["AIS_Model"] = best_model_name

result_df.to_csv(
    output("result.csv"),
    index=False
)

print("\nAIS CLASSIFICATION RESULTS")
display(result_df)

# ============================================================
# 19. AIS RESULT GRAPH
# ============================================================

actual_counts = (
    result_df["Employment_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

predicted_counts = (
    result_df["Predicted_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

result_counts = pd.DataFrame({
    "Actual": actual_counts,
    "AIS Predicted": predicted_counts
})

result_counts.plot(
    kind="bar",
    figsize=(10, 6),
    color=["steelblue", "orange"]
)

plt.title(
    "AIS Actual vs Predicted Employment Distribution",
    fontweight="bold"
)

plt.xlabel("Employment Category")
plt.ylabel("Number of Companies")
plt.xticks(rotation=0)
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("result_graph.png")

# ============================================================
# 20. AIS PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({
    "Company": df["Company"],
    "Sector": df["Sector"],
    "Actual_Total_Employment": df["Total_Employment"],
    "Actual_Category": label_encoder.inverse_transform(y),
    "Predicted_Category": label_encoder.inverse_transform(
        best_predictions
    ),
    "Prediction_Status": np.where(
        y == best_predictions,
        "Correct",
        "Incorrect"
    ),
    "Model": best_model_name
})

prediction_df.to_csv(
    output("prediction.csv"),
    index=False
)

print("\nAIS PREDICTION RESULTS")
display(prediction_df)

# ============================================================
# 21. AIS PREDICTION GRAPH
# ============================================================

ordered_classes = [
    category
    for category in ["Low", "Medium", "High"]
    if category in class_names
]

category_map = {
    name: i
    for i, name in enumerate(ordered_classes)
}

actual_numeric = (
    prediction_df["Actual_Category"]
    .map(category_map)
    .to_numpy()
)

predicted_numeric = (
    prediction_df["Predicted_Category"]
    .map(category_map)
    .to_numpy()
)

indices = np.arange(len(prediction_df))

plt.figure(figsize=(14, 7))

plt.plot(
    indices,
    actual_numeric,
    marker="o",
    linewidth=2,
    label="Actual",
    color="royalblue"
)

plt.plot(
    indices,
    predicted_numeric,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Predicted",
    color="darkorange"
)

plt.xticks(
    indices,
    prediction_df["Company"],
    rotation=75,
    ha="right",
    fontsize=8
)

plt.yticks(
    range(len(ordered_classes)),
    ordered_classes
)

plt.title(
    "AIS Actual vs Predicted Employment Categories",
    fontweight="bold"
)

plt.xlabel("Companies")
plt.ylabel("Employment Category")
plt.legend()
plt.grid(alpha=0.3)

save_show("prediction_graph.png")

# ============================================================
# 22. CLASSIFICATION REPORT
# ============================================================

print("\nAIS CLASSIFICATION REPORT")

print(
    classification_report(
        y,
        best_predictions,
        labels=np.arange(len(class_names)),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 23. TRAIN FINAL AIS-OPTIMIZED MODEL
# ============================================================

final_pipeline = make_pipeline(
    models[best_model_name],
    selected_features
)

final_pipeline.fit(
    X_selected,
    y
)

with open(
    output("shipbuilding_model.pkl"),
    "wb"
) as f:
    pickle.dump(final_pipeline, f)

with open(
    output("preprocessor.pkl"),
    "wb"
) as f:
    pickle.dump(
        final_pipeline.named_steps["preprocessor"],
        f
    )

with open(
    output("label_encoder.pkl"),
    "wb"
) as f:
    pickle.dump(label_encoder, f)

# ============================================================
# 24. TRAIN EXPERIMENTAL NEURAL NETWORK
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

tf.random.set_seed(SEED)

X_nn = (
    final_pipeline.named_steps["preprocessor"]
    .transform(X_selected)
)

X_nn = np.asarray(
    X_nn,
    dtype=np.float32
)

nn_model = keras.Sequential([
    layers.Input(
        shape=(X_nn.shape[1],)
    ),
    layers.Dense(
        32,
        activation="relu"
    ),
    layers.Dropout(0.2),
    layers.Dense(
        16,
        activation="relu"
    ),
    layers.Dense(
        len(class_names),
        activation="softmax"
    )
])

nn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = nn_model.fit(
    X_nn,
    y,
    epochs=80,
    batch_size=4,
    verbose=0
)

nn_model.save(
    str(output("shipbuilding_model.h5"))
)

pd.DataFrame(
    history.history
).to_csv(
    output("neural_network_history.csv"),
    index=False
)

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.title(
    "AIS Feature-Selected Neural Network Training",
    fontweight="bold"
)

plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(alpha=0.3)

save_show("neural_network_accuracy.png")

# ============================================================
# 25. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based Shipbuilding and Ship Repairing "
        "Employment Pattern Analysis"
    ),
    "optimization_algorithm": "AIS Clonal Selection",
    "dataset": DATASET_PATH.name,
    "historical_year": 1996,
    "target": "Employment_Category",
    "selected_features": selected_features,
    "best_model": best_model_name,
    "validation": "Leave-One-Out",
    "ais_parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "clone_factor": CLONE_FACTOR,
        "replacement_rate": REPLACEMENT_RATE
    },
    "thresholds": {
        "low_max": float(q1),
        "medium_max": float(q2)
    },
    "random_state": SEED
}

with open(
    output("model_config.yaml"),
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 26. SAVE JSON METADATA
# ============================================================

best_row = comparison_df.iloc[0]

metadata = {
    "project": config["project"],
    "algorithm": "AIS",
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "selected_features": selected_features,
    "best_fitness": float(best_fitness),
    "best_model": best_model_name,
    "accuracy": float(best_row["Accuracy"]),
    "precision": float(best_row["Precision"]),
    "recall": float(best_row["Recall"]),
    "f1_score": float(best_row["F1_Score"]),
    "neural_network_training_accuracy": float(
        history.history["accuracy"][-1]
    ),
    "evaluation_method": "LOOCV",
    "limitations": [
        "Dataset is very small.",
        "Feature selection used the full dataset.",
        "Reported LOOCV model metrics are not nested-CV estimates.",
        "Feature selection and model selection can cause optimistic scores.",
        "Neural network accuracy is training accuracy only.",
        "This is historical exploratory analysis."
    ]
}

with open(
    output("model_metadata.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4
    )

# ============================================================
# 27. VERIFY ALL AIS OUTPUTS
# ============================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "fitness_graph.png",
    "feature_selection_graph.png",
    "feature_selection.csv",
    "model_comparison.csv",
    "shipbuilding_model.pkl",
    "shipbuilding_model.h5",
    "preprocessor.pkl",
    "label_encoder.pkl",
    "model_config.yaml",
    "model_metadata.json",
    "selected_features.json",
    "neural_network_history.csv",
    "neural_network_accuracy.png"
]

print("\n" + "=" * 70)
print("AIS OUTPUT VERIFICATION")
print("=" * 70)

for filename in output_files:

    path = output(filename)

    if path.exists():

        size_kb = path.stat().st_size / 1024

        print(
            f"[SAVED] ais_{filename} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(f"[MISSING] ais_{filename}")

print("\n" + "=" * 70)
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nBest AIS model:", best_model_name)
print("Selected features:", selected_features)
print("Best AIS fitness:", round(best_fitness, 4))

print("\nAll results saved in:")
print(BASE_DIR)
```
