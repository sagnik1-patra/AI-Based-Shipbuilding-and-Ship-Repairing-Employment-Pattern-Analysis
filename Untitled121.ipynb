```python
import os
import json
import yaml
import pickle
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.model_selection import (
    LeaveOneOut,
    cross_val_predict
)
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Shipbuilding and Ship Repairing Employment Pattern Analysis"
)

CSV_FILE = BASE_DIR / (
    "10.3_Employment_in_Ship_Building_"
    "Ship_Repairing_Companies_As_on_31st_March_1996.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

if not CSV_FILE.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_FILE}"
    )

print("=" * 65)
print("SHIPBUILDING EMPLOYMENT ANALYSIS")
print("=" * 65)
print("Dataset:", CSV_FILE)
print("Output:", BASE_DIR)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(CSV_FILE)

df.columns = df.columns.str.strip()

print("\nOriginal dataset shape:", df.shape)
print("\nDataset columns:")
print(df.columns.tolist())

print("\nFirst five records:")
print(df.head().to_string(index=False))

print("\nMissing values:")
print(df.isnull().sum())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

column_mapping = {
    "S. No.": "Serial_Number",
    "Category of Sectors": "Sector",
    "Name of the Company": "Company",
    "Officers": "Officers",
    "Supervisors": "Supervisors",
    "Workforce - Directly Involved in Operation": "Direct_Workers",
    "Workforce - Indirectly Involved in Operation": "Indirect_Workers",
    "Total": "Total_Employment"
}

df = df.rename(columns=column_mapping)

required_columns = list(column_mapping.values())

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# ============================================================
# 4. CLEAN DATASET
# ============================================================

df["Sector"] = (
    df["Sector"]
    .astype(str)
    .str.strip()
    .replace({
        "A. Public Sector": "Public",
        "B. Private Sector": "Private"
    })
)

df["Company"] = df["Company"].astype(str).str.strip()

numeric_columns = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers",
    "Total_Employment"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col], errors="coerce"
    )

df = df.dropna(subset=["Total_Employment"]).copy()
df = df[df["Total_Employment"] > 0].copy()

# Missing workforce components are retained as missing.
# They are NOT automatically interpreted as zero.
# For descriptive composition analysis only, use available
# component values and a residual "unreported" component.

component_columns = [
    "Officers",
    "Supervisors",
    "Direct_Workers",
    "Indirect_Workers"
]

reported_sum = df[component_columns].fillna(0).sum(axis=1)

df["Unreported_Workforce"] = (
    df["Total_Employment"] - reported_sum
).clip(lower=0)

df["Component_Total_Difference"] = (
    df["Total_Employment"] - reported_sum
)

# ============================================================
# 5. FEATURE ENGINEERING
# ============================================================

# Employment size classes based on fixed tertile thresholds.
# The thresholds are derived from this historical dataset.
# They are NOT a standardized industry classification.

thresholds = np.quantile(
    df["Total_Employment"],
    [1 / 3, 2 / 3]
)

lower_threshold = float(thresholds[0])
upper_threshold = float(thresholds[1])

def employment_category(total):
    if total <= lower_threshold:
        return "Low"
    elif total <= upper_threshold:
        return "Medium"
    return "High"

df["Employment_Category"] = (
    df["Total_Employment"]
    .apply(employment_category)
)

# Use relative composition, not raw workforce counts.
# This reduces direct leakage from Total_Employment.
# Composition still carries information about company size,
# so results must be interpreted as exploratory.

for col in component_columns:
    feature_name = col + "_Share"

    df[feature_name] = (
        df[col] / df["Total_Employment"]
    )

df["Reported_Component_Count"] = (
    df[component_columns].notna().sum(axis=1)
)

# The reported-component count describes data completeness.
# It is excluded from predictors to avoid exploiting
# reporting patterns associated with company size.

feature_columns = [
    "Sector",
    "Officers_Share",
    "Supervisors_Share",
    "Direct_Workers_Share",
    "Indirect_Workers_Share"
]

X = df[feature_columns].copy()
y_text = df["Employment_Category"].copy()

label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_text)

class_names = list(label_encoder.classes_)

print("\nEmployment category thresholds:")
print("Low <=", lower_threshold)
print("Medium <=", upper_threshold)
print("High >", upper_threshold)

print("\nCategory distribution:")
print(df["Employment_Category"].value_counts())

print("\nFeatures:")
print(feature_columns)

# Save processed data
df.to_csv(
    BASE_DIR / "processed_dataset.csv",
    index=False
)

# ============================================================
# 6. PREPROCESSING PIPELINE
# ============================================================

numeric_features = [
    "Officers_Share",
    "Supervisors_Share",
    "Direct_Workers_Share",
    "Indirect_Workers_Share"
]

categorical_features = ["Sector"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    )
])

def make_preprocessor():
    return ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ])

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=3,
        random_state=SEED
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=3,
        random_state=SEED
    ),

    "Support Vector Machine": SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale"
    ),

    "K-Nearest Neighbors": KNeighborsClassifier(
        n_neighbors=3
    )
}

# ============================================================
# 8. LEAVE-ONE-OUT CROSS VALIDATION
# ============================================================

# With only 18 records, LOOCV is more appropriate
# than relying on a single small train/test split.
#
# Preprocessing is fitted inside each training fold,
# avoiding imputation/scaling leakage.
#
# Category thresholds are fixed from the full dataset
# for this descriptive classification experiment.

loo = LeaveOneOut()

results = []
predictions_by_model = {}

print("\n" + "=" * 65)
print("MACHINE LEARNING MODEL EVALUATION")
print("=" * 65)

for model_name, estimator in models.items():

    print(f"\nTraining: {model_name}")

    pipeline = Pipeline([
        ("preprocessor", make_preprocessor()),
        ("classifier", estimator)
    ])

    y_pred = cross_val_predict(
        pipeline,
        X,
        y,
        cv=loo,
        n_jobs=1
    )

    accuracy = accuracy_score(y, y_pred)

    precision = precision_score(
        y, y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y, y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y, y_pred,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    predictions_by_model[model_name] = y_pred

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")

# ============================================================
# 9. MODEL COMPARISON
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    BASE_DIR / "model_comparison.csv",
    index=False
)

print("\nMODEL COMPARISON:")
print(comparison_df.to_string(index=False))

best_model_name = comparison_df.iloc[0]["Model"]

print("\nBest model:", best_model_name)

# ============================================================
# 10. TRAIN FINAL SCIKIT-LEARN MODEL
# ============================================================

best_pipeline = Pipeline([
    ("preprocessor", make_preprocessor()),
    ("classifier", models[best_model_name])
])

best_pipeline.fit(X, y)

# Save final model as PKL
with open(
    BASE_DIR / "shipbuilding_model.pkl", "wb"
) as f:
    pickle.dump(best_pipeline, f)

# Save fitted preprocessor separately
fitted_preprocessor = (
    best_pipeline.named_steps["preprocessor"]
)

with open(
    BASE_DIR / "preprocessor.pkl", "wb"
) as f:
    pickle.dump(fitted_preprocessor, f)

# Save label encoder
with open(
    BASE_DIR / "label_encoder.pkl", "wb"
) as f:
    pickle.dump(label_encoder, f)

print("\nSaved sklearn model and preprocessing files.")

# ============================================================
# 11. NEURAL NETWORK MODEL
# ============================================================

# Neural network is trained as an additional experimental
# model. Its training accuracy is NOT a valid estimate
# of generalization performance.

X_nn = fitted_preprocessor.transform(X)
X_nn = np.asarray(X_nn, dtype=np.float32)

num_features = X_nn.shape[1]
num_classes = len(class_names)

nn_model = keras.Sequential([
    layers.Input(shape=(num_features,)),

    layers.Dense(
        32,
        activation="relu"
    ),

    layers.Dropout(0.2),

    layers.Dense(
        16,
        activation="relu"
    ),

    layers.Dense(
        num_classes,
        activation="softmax"
    )
])

nn_model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nNeural Network Architecture:")
nn_model.summary()

history = nn_model.fit(
    X_nn,
    y,
    epochs=80,
    batch_size=4,
    verbose=0,
    shuffle=True
)

# Save trained Keras model in requested H5 format
nn_model.save(
    str(BASE_DIR / "shipbuilding_model.h5")
)

history_df = pd.DataFrame(history.history)

history_df.to_csv(
    BASE_DIR / "neural_network_history.csv",
    index=False
)

print("\nNeural network saved successfully.")

# ============================================================
# 12. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project_name": (
        "AI-Based Shipbuilding and Ship Repairing "
        "Employment Pattern Analysis"
    ),

    "dataset": CSV_FILE.name,

    "historical_year": 1996,

    "task": "Employment Size Classification",

    "target_column": "Employment_Category",

    "feature_columns": feature_columns,

    "employment_thresholds": {
        "low_max": lower_threshold,
        "medium_max": upper_threshold
    },

    "classes": class_names,

    "validation": "Leave-One-Out Cross Validation",

    "best_model": best_model_name,

    "random_seed": SEED,

    "model_files": {
        "sklearn": "shipbuilding_model.pkl",
        "keras": "shipbuilding_model.h5",
        "preprocessor": "preprocessor.pkl",
        "label_encoder": "label_encoder.pkl"
    }
}

with open(
    BASE_DIR / "model_config.yaml",
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("YAML configuration saved.")

# ============================================================
# 13. SAVE JSON METADATA
# ============================================================

best_row = comparison_df.iloc[0]

metadata = {
    "project_name": config["project_name"],

    "created_at": datetime.now().isoformat(),

    "dataset_name": CSV_FILE.name,

    "total_records": int(len(df)),

    "total_features": int(len(feature_columns)),

    "historical_year": 1996,

    "sector_distribution": {
        str(k): int(v)
        for k, v in df["Sector"].value_counts().items()
    },

    "employment_categories": {
        str(k): int(v)
        for k, v in (
            df["Employment_Category"]
            .value_counts()
            .items()
        )
    },

    "thresholds": {
        "low_max": lower_threshold,
        "medium_max": upper_threshold
    },

    "best_model": best_model_name,

    "best_model_metrics": {
        "accuracy": float(best_row["Accuracy"]),
        "precision": float(best_row["Precision"]),
        "recall": float(best_row["Recall"]),
        "f1_score": float(best_row["F1_Score"])
    },

    "validation_method": "Leave-One-Out",

    "neural_network_training_accuracy": float(
        history.history["accuracy"][-1]
    ),

    "models_evaluated": list(models.keys()),

    "feature_columns": feature_columns,

    "class_names": class_names,

    "limitations": [
        "Dataset contains only 18 historical records.",
        "Model results are exploratory, not production-grade.",
        "Neural network accuracy is training accuracy only.",
        "Employment categories use dataset-derived thresholds.",
        "Missing workforce values are not assumed to be zero."
    ]
}

with open(
    BASE_DIR / "model_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4,
        ensure_ascii=False
    )

print("JSON metadata saved.")

# ============================================================
# 14. CLASSIFICATION RESULTS
# ============================================================

best_cv_predictions = predictions_by_model[
    best_model_name
]

classification_results = df[[
    "Company",
    "Sector",
    "Total_Employment",
    "Employment_Category"
]].copy()

classification_results["Predicted_Category"] = (
    label_encoder.inverse_transform(
        best_cv_predictions
    )
)

classification_results["Correct_Prediction"] = (
    classification_results["Employment_Category"]
    ==
    classification_results["Predicted_Category"]
)

classification_results.to_csv(
    BASE_DIR / "classification_results.csv",
    index=False
)

print("\nClassification Report:")
print(
    classification_report(
        y,
        best_cv_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 15. MODEL ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=comparison_df,
    x="Accuracy",
    y="Model",
    color="steelblue"
)

plt.title("Machine Learning Model Accuracy Comparison")
plt.xlabel("Leave-One-Out Accuracy")
plt.ylabel("Model")
plt.xlim(0, 1)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 16. CONFUSION MATRIX HEATMAP
# ============================================================

cm = confusion_matrix(
    y,
    best_cv_predictions,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(f"Confusion Matrix - {best_model_name}")
plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 17. MODEL COMPARISON GRAPH
# ============================================================

comparison_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1_Score"]
].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Machine Learning Model Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=30, ha="right")
plt.legend(loc="best")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 18. NEURAL NETWORK TRAINING GRAPH
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Neural Network Training Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "neural_network_accuracy.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. VERIFY SAVED FILES
# ============================================================

required_files = [
    "shipbuilding_model.h5",
    "shipbuilding_model.pkl",
    "preprocessor.pkl",
    "label_encoder.pkl",
    "model_config.yaml",
    "model_metadata.json",
    "model_comparison.csv",
    "processed_dataset.csv",
    "classification_results.csv",
    "neural_network_history.csv",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "neural_network_accuracy.png"
]

print("\n" + "=" * 65)
print("FINAL OUTPUT VERIFICATION")
print("=" * 65)

for filename in required_files:

    filepath = BASE_DIR / filename

    if filepath.exists():
        size_kb = filepath.stat().st_size / 1024

        print(
            f"[SAVED] {filename} "
            f"({size_kb:.2f} KB)"
        )
    else:
        print(f"[MISSING] {filename}")

print("\nProject completed successfully!")
print("All outputs saved in:")
print(BASE_DIR)
```
